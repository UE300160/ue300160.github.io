### try 1:

* reverse the DNA string
* go over each reverse position, translate to complementary

```python
dna = "AAAACCCGGT"
reverse = dna[::-1]
rev_compl = ""
for c in reverse:
    if c == "A":
        rev_compl += "T"
    elif c == "T":
        rev_compl += "A"
    elif c == "C":
        rev_compl += "G"
    elif c == "G":
        rev_compl += "C"

print(rev_compl)
```

## try 2: using a dictionary

* reverse the DNA
* instead of asking "if" each time, we save the complements in a dictionary and just refer to that.

```python
dna = "AAAACCCGGT"
complements = {"A": "T", "T": "A", "C": "G", "G": "C"}
reverse = dna[::-1]
rev_compl = ""
for c in reverse:
    rev_compl += complements[c]

print(rev_compl)
```

## try 3: lose the reversing

* ~reverse the DNA~
* instead of asking if use a dictionary

```python
dna = "AAAACCCGGT"
complements = {"A": "T", "T": "A", "C": "G", "G": "C"}
rev_compl = ""
for c in dna: # we could also have changed the way we read the dna[::-1]
    rev_compl = complements[c] + rev_compl # reverse the way we build up the accumulator

print(rev_compl)
```

```python
# 1. read the input file, save as a list
# 2. keep even-numbered lines
# 3. write a file with the even-numbered lines

# 1. read and save as a list
input_file = []
with open(file_location, "r") as f:
    for line in f:
        input_file.append(line)
```

### brief aside

`enumerate` is how we get access to indices and values at the same time. If we find ourselves writing `range(len` then we need enumerate.

```python
test = "abcdefg"
# for i, c in enumerate(test):
for i in range(len(test)):
    c = test[i]
    print(i, c)
```

```python
# 2. keep even-numbered lines, i.e. odd-numbered indices!
# even_lines = input_file[1::2]

even_lines = []
for i, line in enumerate(input_file):
    if i % 2 == 1:
        even_lines.append(line)
```

```python
even_lines
```

```python
# 3. write the even lines in a new file
output_location = "./even_lines.txt"
with open(output_location, "w") as f:
    for line in even_lines:
        f.write(line)
```
