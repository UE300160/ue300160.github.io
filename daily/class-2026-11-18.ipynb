```python
mass = 3.54
if mass > 3.0:
    print("this is too heavy")

if mass <= 4.0:
    print("this fits in the box")

if mass == 3.54:
    print("this is weirdly specific")
```

```python
masses = [3.54, 5, 7, 1, 0.4, 2.68]

for m in masses:
    if m > 3.0:
        print(m, "is too heavy")
```

Given a list of masses, calculate how much I will carry in total if I only carry items weighing less than 3.

```python
masses = [3.54, 5, 7, 1, 0.4, 2.68, 8, 12, 5, 1, 2, 0.2, 0.3]

total_weight = 0
for m in masses:
    if m < 3.0:
        # put it in the trunk
        total_weight = total_weight + m
print("my trunk contains items with a total mass of", total_weight)
```

## use `else` to do something when the condition is _not_ true

```python
for m in masses:
    if m > 3.0:
        print(m, "is too heavy")
    else:
        print(m, "is very light")
```

```python
lights = ["green", "red", "yellow", "blueeeee", "green", "green", "red", "yellow", "green", "green", "red", "yellow", "green"]

for l in lights:
    if l == "green":
        print("keep moving")
    elif l == "yellow":
        print("FASTER")
    elif l == "red":
        print("stop")
    else:
        print("maybe stop?")
```

## more complex decision-making

Given a list of masses, calculate how much I will carry in total if I only carry items weighing less than 6 and more than 3.

```python
masses = [3.54, 5, 7, 1, 0.4, 2.68, 8, 12, 5, 1, 2, 0.2, 0.3]

total_weight = 0
for m in masses:
    if m < 6.0 and m > 3.0:
            # put it in the trunk
            total_weight = total_weight + m
print("my trunk contains items with a total mass of", total_weight)
```

Given a list of masses, calculate how much I will carry in total if I only carry items weighing less than 3 or more than 10.

```python
masses = [3.54, 5, 7, 1, 0.4, 2.68, 8, 12, 5, 1, 2, 0.2, 0.3]

total_weight = 0
for m in masses:
    if m < 3.0 or m > 10.0:
            # put it in the trunk
            total_weight = total_weight + m
print("my trunk contains items with a total mass of", total_weight)
```
