```python
def gc(dna):
    no_C = dna.count("C")
    no_G = dna.count("G")
    length = len(dna)
    return (no_C + no_G) / length
```

```python
sequences = {}

with open("./GC_test.txt", "r") as rosalind:
    for line in rosalind:
        # figure out if it's a header or a sequence
        if line[0] == ">":
            # is a header!
            last_id = line[1:].rstrip()
            sequences[last_id] = ""
        else: # this is a sequence line!
            sequences[last_id] += line.rstrip()
```

```python
gc_content = {}
for seq_id, seq in sequences.items():
    gc_content[seq_id] = gc(seq)
```

```python
max_gc = max(gc_content.values())

for seq_id in gc_content:
    if gc_content[seq_id] == max_gc:
        print(seq_id)
        print(round(max_gc, 4))
```
