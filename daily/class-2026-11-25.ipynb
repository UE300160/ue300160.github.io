```python
n = 10
s = 'et cetera'
print(s)
i = 0
while i < n:
    # print('at', j)
    new = ''
    for j in range(len(s)):
        left = j-1
        right = (j+1)%len(s)
        if s[left]==s[right]: new = new + '-'
        else: new = new + '*'
    s=''.join(new)
    print(s)
    i += 1
```

```python
n = 50
s = 'efefefefefefefefefefefe'
print(s)
for i in range(n):
    # print('at', j)
    new = '' # define an empty string
    for j, c in enumerate(s): # go over all positions of the string
        # look to the positions left and right of j
        left = j - 1
        right = (j + 1) % len(s)
        if s[left] == s[right]: # if equal, write -
            new = new + '-'
        else: # else write *
            new = new + '*'
    s = ''.join(new) # replace the input string
    print(s)
```
