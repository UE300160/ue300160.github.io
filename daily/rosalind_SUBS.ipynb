## SUBS

important to remember: Rosalind "location" is human-readable, so `index + 1`.

#### Sample Dataset

```
GATATATGCATATACTT
ATAT
```

#### Sample Output

```
2 4 10
```

````python
s = "GATATATGCATATACTT"
t = "ATAT"
````

````python
# let's print all substrings of length len(t)
steps = len(s) - len(t) + 1
sub_length = len(t)
substrings = []
for i in range(steps):
    substrings.append(s[i:i+sub_length])
````
